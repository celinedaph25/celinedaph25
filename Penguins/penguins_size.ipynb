```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

%matplotlib inline

penguins = pd.read_csv('penguins_size.csv', keep_default_na=True, na_values=['NA', 'NaN', 'nan', '.'])
penguins.info()
```

### Male vs Female beak sizes

```python
plt.figure()
fg = sns.FacetGrid(penguins, row='sex', col = 'island', hue='species', margin_titles=True)
fg.map(sns.scatterplot, 'culmen_length_mm', 'culmen_depth_mm')
fg.add_legend()
plt.show()
```

### Body mass vs flipper length

```python
plt.figure()
fg = sns.FacetGrid(penguins, row='sex', col = 'island', hue='species', margin_titles=True)
fg.map(sns.scatterplot, 'body_mass_g', 'flipper_length_mm')
fg.add_legend()
plt.show()
```
